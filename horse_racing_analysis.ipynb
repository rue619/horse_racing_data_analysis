```python
horse_race_data["rating_band"] = pd.cut(
    horse_race_data["horse_rating"],
    bins=[0, 39, 59, 79, 99, float("inf")],
    labels=["0-39", "40-59", "60-79", "80-99", "100+"]
)

rating_summary = horse_race_data.groupby(
    "rating_band",
    observed=True
).agg(
    horse_runs=("horse_id", "count"),
    average_finish=("result", "mean"),
    top_3_rate=("top_3", "mean")
).reset_index()
rating_summary["top_3_rate_pct"] = rating_summary["top_3_rate"] * 100
rating_summary
```

```python
horse_race_data["field_size_group"] = pd.cut(
    horse_race_data["field_size"],
    bins=[0, 11, 14],
    labels=["4-11 horses", "12-14 horses"]
)

draw_field_summary = horse_race_data.groupby(
    ["field_size_group", "draw"],
    observed=True
).agg(
    horse_runs=("horse_id", "count"),
    top_3_rate=("top_3", "mean")
).reset_index()

draw_field_summary
```

```python
horse_race_data["odds_band"] = pd.cut(
    horse_race_data["win_odds"],
    bins=[0, 5, 10, 20, 50, float("inf")],
    labels=["1-5", "5-10", "10-20", "20-50", "50+"]
)

odds_summary = horse_race_data.groupby(
    "odds_band",
    observed=True
).agg(
    horse_runs=("horse_id", "count"),
    average_finish=("result", "mean"),
    top_3_rate=("top_3", "mean"),
    win_rate=("won", "mean")
).reset_index()
odds_summary["win_rate_pct"] = odds_summary["win_rate"] * 100
odds_summary
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.bar(
    odds_summary["odds_band"],
    odds_summary["win_rate_pct"]
)

plt.title("Win Rate by Win-Odds Band")
plt.xlabel("Win Odds")
plt.ylabel("Win Rate (%)")

plt.savefig(
    "visualizations/win_rate_by_odds.png",
    bbox_inches="tight"
)

plt.show()
```

# Key Findings & Insights

## 1. Horse Rating and Performance

Horse rating was associated with race performance. Horses in the 100+ rating group had the best average finishing position and the highest Top-3 rate among the rating groups analyzed.

The relationship was not perfectly linear across all rating bands, so higher rating should be interpreted as an indicator associated with performance rather than a guarantee of success.

## 2. Starting Odds and Performance

Win odds showed a strong relationship with race performance. Horses with lower odds generally had better average finishing positions and higher win and Top-3 rates.

For example, horses with odds between 1 and 5 had a substantially higher Top-3 rate than horses with odds above 50.

This suggests that the market's expectations were strongly associated with observed race outcomes in this dataset.

## 3. Draw Position and Performance

Lower draw numbers generally showed better finishing performance.

Draws 1–3 had higher Top-3 rates and better average finishing positions than many of the higher draws. Performance generally became weaker as draw number increased, although the relationship was not perfectly consistent.

## 4. Race Conditions

Performance varied across race conditions, including distance, venue, track going, and field size.

These differences suggest that race conditions are useful dimensions for further analysis when evaluating horse performance.

## 5. Overall Dataset

The analysis used 6,348 races and 79,447 horse-run records.

The dashboard was created to explore how horse characteristics, starting odds, draw position, and race conditions were associated with race performance.
